# 4. Understand Common Dataset Formats

| Format | Typical extension | Pandas loading function | Common use |
|---|---|---|---|
| CSV | `.csv` | `pd.read_csv()` | Tabular data exchange |
| Excel | `.xlsx`, `.xls` | `pd.read_excel()` | Business reports and spreadsheets |
| JSON | `.json` | `pd.read_json()` | Nested or API-style records |
| Text | `.txt` | `pd.read_csv()` | Delimited text files |
| Parquet | `.parquet` | `pd.read_parquet()` | Efficient analytical storage |
| SQL database | Database table | `pd.read_sql()` | Structured operational/analytical data |
| Web URL | URL | Often `pd.read_csv()` | Publicly hosted files |
| Scikit-learn dataset | Built-in object | Dataset loader + DataFrame | Learning and benchmarking |

**Important:** The correct loading function depends on the actual file structure. A file extension alone does not guarantee that the contents are valid or correctly formatted.

**Source:** UCI Machine Learning Repository — Iris Dataset. The raw file contains 150 records, four measurements, and a species label. The source file has no header row, so we will provide column names while reading it; we will not alter the original raw file.

If you already downloaded `iris.data`, place it directly inside `data/raw/`. Otherwise, run the next cell while connected to the internet.

````python
from pathlib import Path
from urllib.request import urlretrieve
import pandas as pd

# Current notebook working directory
WORKING_DIR = Path.cwd()

# Notebook project folder ke parent ko Task-01 folder maana jayega
TASK_DIR = WORKING_DIR.parent

# Raw dataset location
RAW_DIR = TASK_DIR / "data" / "raw"
RAW_FILE = RAW_DIR / "iris.data"

# Create raw directory
RAW_DIR.mkdir(parents=True, exist_ok=True)

# Official UCI Iris dataset URL
UCI_IRIS_URL = (
    "https://archive.ics.uci.edu/"
    "ml/machine-learning-databases/iris/iris.data"
)

# Download only if file does not exist
if not RAW_FILE.exists():
    urlretrieve(UCI_IRIS_URL, RAW_FILE)
    print("Iris dataset downloaded successfully!")
else:
    print("Using existing Iris raw dataset.")

# Column names
column_names = [
    "sepal_length_cm",
    "sepal_width_cm",
    "petal_length_cm",
    "petal_width_cm",
    "species"
]

# Load raw dataset
df = pd.read_csv(
    RAW_FILE,
    header=None,
    names=column_names
)

# Display dataset information
print("Task directory:", TASK_DIR.resolve())
print("Dataset location:", RAW_FILE.resolve())
print("Dataset shape:", df.shape)

display(df.head())
````

# 7. Initial Data Inspection

After loading a dataset, inspect it before performing transformations. These checks help establish what is present in the data and identify issues that may affect later analysis.

## 7.1 Preview the first and last rows

- `head()` displays the first rows.
- `tail()` displays the last rows.
- `sample()` displays randomly selected rows.

These previews help detect unexpected headers, formatting problems, or unusual records.

````python
display(df.head())
display(df.tail())
display(df.sample(5, random_state=42))
````

## 7.2 Inspect dimensions and structure

`shape` gives the row and column count. `info()` provides non-null counts, data types, and memory information.

````python
print("Shape:", df.shape)
print("\nDataset information:")
df.info()
````

## 7.3 Inspect column names and data types

Data types matter because numeric measurements should generally be numeric, while labels such as species should be treated as categories.

````python
print("Columns:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes)

print("\nData type summary:")
display(df.dtypes.value_counts())
````

## 7.4 Check missing values

Missing values can affect summaries and later model preparation. First measure them; do not automatically delete rows without understanding why values are missing.

````python
missing_count = df.isna().sum()
missing_percent = (df.isna().mean() * 100).round(2)

missing_report = pd.DataFrame({
    "missing_count": missing_count,
    "missing_percent": missing_percent
})

display(missing_report)
````

## 7.5 Check duplicate rows

Duplicate records can affect analysis, but a duplicate should not be removed automatically. First inspect how many exist and determine whether they represent accidental duplication or legitimate repeated measurements.

````python
duplicate_count = df.duplicated().sum()
print("Number of fully duplicated rows:", duplicate_count)

# Display duplicate rows for review (if any)
display(df[df.duplicated(keep=False)].head(10))
````

## 7.6 Summary statistics

`describe()` summarises numeric columns using count, mean, standard deviation, minimum, quartiles, and maximum. These values help identify ranges that may need further investigation.

````python
display(df.describe().T)
````

## 7.7 Inspect categorical values

For the target column, check the distinct labels and the number of observations in each class.

````python
print("Unique species:")
print(df["species"].unique())

print("\nSpecies count:")
display(df["species"].value_counts())

print("\nSpecies proportions:")
display(df["species"].value_counts(normalize=True).mul(100).round(2))
````

## 7.8 Check memory usage

Memory usage can become important when datasets grow. The result below reports the approximate memory used by each column and the total.

````python
memory_report = df.memory_usage(deep=True).to_frame(name="memory_bytes")
display(memory_report)

total_memory_kb = df.memory_usage(deep=True).sum() / 1024
print(f"Approximate DataFrame memory: {total_memory_kb:.2f} KB")
````

# 8. Initial Data Quality Review

Use the inspection results to create a concise quality checklist.

| Check | What to look for | Why it matters |
|---|---|---|
| Row and column count | Expected dataset size and fields | Confirms the data loaded as expected |
| Column names | Clear, meaningful names | Prevents confusion in later steps |
| Data types | Numeric measurements and categorical label | Supports correct calculations |
| Missing values | Count and percentage per column | Missing data may affect analysis |
| Duplicate rows | Repeated complete records | May indicate repeated or duplicated observations |
| Numeric ranges | Unexpected minimums or maximums | May reveal invalid values or unusual observations |
| Class distribution | Counts for each species | Shows whether classes are represented evenly |

**Interpretation rule:** A check identifies something to investigate; it does not automatically prove that the data is wrong. For example, a repeated row may be a valid observation.

````python
# Compact inspection summary
inspection_summary = {
    "rows": df.shape[0],
    "columns": df.shape[1],
    "missing_cells": int(df.isna().sum().sum()),
    "duplicate_rows": int(df.duplicated().sum()),
    "numeric_columns": int(df.select_dtypes(include=np.number).shape[1]),
    "categorical_columns": int(df.select_dtypes(exclude=np.number).shape[1])
}

display(pd.Series(inspection_summary, name="value").to_frame())
````
